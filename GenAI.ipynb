```python
# 2. Import the libraries we need
from transformers import pipeline, GPT2Tokenizer, GPT2LMHeadModel
import torch

print("Libraries imported successfully!")
```

```python
# 3. Specify the model name
model_name = "gpt2"

# 4. Load the tokenizer
# The tokenizer converts text to numbers (tokens) and back
tokenizer = GPT2Tokenizer.from_pretrained(model_name)

# 5. Load the model
# This downloads the pre-trained GPT-2 model
model = GPT2LMHeadModel.from_pretrained(model_name)

print("Model and tokenizer loaded successfully!")
print(f"Vocabulary size: {len(tokenizer)} tokens")
```

```python
# If you get a warning about pad_token_id, run this:
tokenizer.pad_token = tokenizer.eos_token
```

```python
# 6. Tokenization demo
text = "ChatGPT is amazing! We are listening to Glowlogics Internship Training"

# Convert text to tokens (numbers)
tokens = tokenizer.encode(text)

print(f"Original text: {text}")
print(f"Tokens (numbers): {tokens}")

# Convert tokens back to text
decoded = tokenizer.decode(tokens)
print(f"Decoded text: {decoded}")

# Show how each token maps to text
print("\n--- Token Breakdown ---")
for token_id in tokens:
    token_text = tokenizer.decode([token_id])
    print(f"Token {token_id}: '{token_text}'")
```

```python
# 7. Compare different texts
texts = [
    "Hello world!",
    "The quick brown fox jumps over the lazy dog.",
    "Machine learning is fascinating."
]

for text in texts:
    tokens = tokenizer.encode(text)
    print(f"Text: {text}")
    print(f"Tokens: {tokens}")
    print(f"Number of tokens: {len(tokens)}")
    print()
```

```python
# 8. Function to generate text
def generate_text(prompt, max_length=50, temperature=0.8):
    """
    Generate text using GPT-2.

    Parameters:
    - prompt: The starting text
    - max_length: Maximum number of tokens to generate
    - temperature: Controls randomness (0 = deterministic, 1 = random)
    """
    # Step 1: Convert the prompt to tokens
    # 'return_tensors="pt"' means return PyTorch tensors
    input_ids = tokenizer.encode(prompt, return_tensors='pt')

    # Step 2: Generate text
    output = model.generate(
        input_ids,               # The input tokens
        max_length=max_length,   # Maximum length of output
        temperature=temperature, # Creativity dial
        do_sample=True,          # Sample from probabilities
        pad_token_id=tokenizer.eos_token_id  # End of sequence token
    )

    # Step 3: Convert tokens back to text
    return tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
# 14. Using the pipeline API (easier way)
from transformers import pipeline

# Create a text generation pipeline
# This loads the model and tokenizer automatically
generator = pipeline('text-generation', model='gpt2')

# Generate text
prompt = "The meaning of life is"
result = generator(prompt, max_length=50, temperature=0.7)

print(f"Prompt: {prompt}")
print(f"Generated: {result[0]['generated_text']}")
```

```python
# 16. Demonstration of limitations
print("--- DEMONSTRATING LIMITATIONS ---\n")

# The model might generate incorrect facts
prompt = "The capital of France is"
print(f"Prompt: {prompt}")
print(f"Generated: {generate_text(prompt, max_length=20, temperature=0.3)}")

# The model might contradict itself
prompt = "The best way to lose weight is"
print(f"\nPrompt: {prompt}")
print(f"Generated: {generate_text(prompt, max_length=50, temperature=0.8)}")

# The model might generate the same thing differently
print("\nSame prompt, different generations:")
for i in range(3):
    generated = generate_text("The meaning of life is", max_length=20, temperature=0.9)
    print(f"  {i+1}. {generated}")
```
