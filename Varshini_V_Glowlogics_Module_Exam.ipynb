# Feature Engineering

## Create Movie Performance Category

Teacher Says:

```text
Rating Below 6
=
Poor

Rating 6 - 8
=
Average

Rating Above 8
=
Excellent
```

### Create Function

````python
def performance_category(rating):
    if rating < 6:
        return 'Poor'
    elif rating >= 6 and rating <= 8:
        return 'Average'
    else:
        return 'Excellent'
````

### Create Performance_Category Column

````python
df["Performance_Category"] = df["IMDB_Rating"].apply(performance_category)
````

### Count Performance Categories

````python
df[["Movie_ID", "IMDB_Rating", "Performance_Category"]].groupby("Performance_Category").count()
````
