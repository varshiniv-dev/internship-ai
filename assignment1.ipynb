```python
# bonus3 - 3 attempts to enter correct atm pin
PIN = 1234
attempts = 3
while attempts > 0:
    pin = int(input("Enter 4-digit PIN: "))
    if pin == PIN:
        print("Transaction allowed!")
        break
    else:
        attempts -= 1
        if attempts > 0:
            print(f"Incorrect PIN. you have {attempts} attempts left.")
if attempts == 0:
    print("Card blocked")
```
