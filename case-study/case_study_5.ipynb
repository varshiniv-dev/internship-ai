```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/health_fitness_dataset.csv")
df.head()
```

```python
print("Shape of dataset: ", df.shape) #display the num of rows and cols
print("\nColumns names: ")
print(df.columns)
print("Dataset information: ")
print(df.info())
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Duplicate rows: ", df.duplicated().sum())
```

```python
#calculate the avg daily steps accross all users
avg_steps = df['Daily_Steps'].mean()
print("Average daily steps: ", avg_steps)
```

```python
#filter user who walk more than 10,000 steps per day
power_users = df[df['Daily_Steps'] > 10000]
print("Power users: ")
print(power_users)
print("number of ower users: ", len(power_users))
```

```python
#create a histogran showing the distribution of daily steps
plt.figure(figsize=(8,6))
sns.histplot(df['Daily_Steps'], bins=20, kde=True)
plt.title('Distribution of Daily Steps')
plt.xlabel('Daily Steps')
plt.ylabel('Frequency')
plt.show()
```

```python
#create a scatter plot showing the relationship between height and weight
plt.figure(figsize=(8,6))
sns.scatterplot(x='Height_cm', y='Weight_kg', data=df)
plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
#create a new column called active_use,  users with at least 8,000 daily steps are considered active
df["Active_User"] = df["Daily_Steps"] >= 8000
df.head()
```

```python
#count active and inactive users
activity_counts = df["Active_User"].value_counts()
print("Activity counts: ")
print(activity_counts)
```
