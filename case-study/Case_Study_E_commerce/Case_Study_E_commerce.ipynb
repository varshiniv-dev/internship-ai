```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
total_orders = len(df)
print("Total orders: ", total_orders)
```

```python
num_of_categories = df["Product_Category"].nunique()
print("Number of categories: ", num_of_categories)
```

```python
categories = df["Product_Category"].unique()
print("Categories: ", categories)
```

```python
average_price = df["Price"].mean()
print("Average price: ", average_price)
```

```python
highest_price = df["Price"].max()
print("Highest price: ", highest_price)
```

```python
premium_rated_products = df[df["Price"]>3000]
print("Premium rated products: ", premium_rated_products)
```

```python
highly_rated_products = df[df["Customer_Rating"]>4]
print("Highly rated products: ", highly_rated_products)
```

```python
electonic_products = df[df["Product_Category"]=="Electronics"]
print("Electronic products: ", electonic_products)
```

```python
clothing_orders = len(df[df["Product_Category"]=="Clothing"])
print("Clothing orders: ", clothing_orders)
```

```python
expensive_electronics = df[(df["Product_Category"]=="Electronics") & (df["Price"]>3000)]
print("Expensive electronics: ", expensive_electronics)
```

```python
lowest_price_first = df.sort_values("Price")
print( lowest_price_first)
```

```python
highest_price_first = df.sort_values("Customer_Rating", ascending=False)
print(highest_price_first)
```

```python
#revenue calculation
df["Revenue"] = df["Price"] * df["Quantity"]
total_revenue = df["Revenue"].sum()
print("Total revenue: ", total_revenue)
```

```python
#average order value
average_order_value = total_revenue / total_orders
print("Average order value: ", average_order_value)
```

```python
#identify top perforn=mance
top_performing_category = df.groupby("Product_Category")["Revenue"].sum().idxmax()
print("Top performing category: ", top_performing_category)
top_category = df.groupby("Product_Category")["Quantity"].sum().idxmax()
print("Top category: ", top_category)
revenue_by_category = df.groupby("Product_Category")["Revenue"].sum()
print("Revenue by category: ", revenue_by_category)
```

```python
#customer satisfaction
avg_rating = df["Customer_Rating"].mean()
print("Avg customer rating: ", round(avg_rating, 2))
```

```python
#premium / highly rated products
premium_rated_products = df[df["Price"]>3000]
print("Premium rated products: ", premium_rated_products)
highly_rated_products = df[df["Customer_Rating"]>4]
print("Highly rated products: ", highly_rated_products)
```

```python
#bar chart
plt.figure(figsize=(10,6))
plt.bar(df["Product_Category"], df["Revenue"])
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.title("Total Revenue by Product Category")
plt.xticks(rotation=45)
plt.show()
```

```python
#price histogram
plt.figure(figsize=(8,6))
plt.hist(df["Price"], bins=20, color="skyblue", edgecolor="black")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Distribution of Product Prices")
plt.show()
```
