```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("/content/student_performance_dataset.csv")
df
```

```python
# 1. Average Marks
average_marks = df["Marks"].mean()
print(f"Average Marks: {average_marks:.2f}")
```

```python
# 2. High Performers (Marks > 75)
high_performers = df[df["Marks"] > 75]
print("Number of students scoring above 75:", len(high_performers))
```

```python
# 3. Correlation Analysis
correlation = df["Attendance_Percentage"].corr(df["Marks"])
print(f"Correlation Coefficient: {correlation:.3f}")
if correlation > 0:
    print("Interpretation: Higher attendance is associated with higher marks.")
elif correlation < 0:
    print("Interpretation: Higher attendance is associated with lower marks.")
else:
    print("Interpretation: No relationship between attendance and marks.")
```

```python
# 4. Feature Engineering
df["Pass"] = df["Marks"] >= 50
```

```python
# 5. Summary Statistics
print("\nPass Count:", df["Pass"].sum())
print("Fail Count:", (~df["Pass"]).sum())
```

```python
# 6. Histogram
plt.figure(figsize=(8,5))
sns.histplot(df["Marks"], bins=10, kde=True)
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(True)
plt.show()
```
