```python
df["Total_Sales"] = (
 df["Price"] * df["Quantity"]
)
print(df["Total_Sales"])
```

```python
df["Total_Sales"].mean() #avg revenue per order
```

```python
category_sales = (
 df.groupby("Product_Category")
 ["Total_Sales"]
 .sum()
 .sort_values(ascending=False)
)

print(category_sales) #Total revenue by category
```

```python
top_category = category_sales.idxmax()
print(f"the highest revenue category is: {top_category}")
```

```python
def recommend_action(row):
 if (
  row["Total_Sales"] > 10000
  and
  row["Customer_Rating"] > 4
 ):
  return "PROMOTE HEAVILY"
 elif (
  row["Total_Sales"] < 5000
  and
  row["Customer_Rating"] < 3
 ):
  return "CONSIDER DISCONTINUE"
 else:
   return "KEEP AS IS"
df.apply(recommend_action, axis=1)
```

```python
df["Business_Action"] = (
 df.apply(
 recommend_action,
 axis=1
 )
 )

df[
 [
 "Order_ID",
 "Product_Category",
 "Total_Sales",
 "Customer_Rating",
 "Business_Action"
 ]
 ]
```
